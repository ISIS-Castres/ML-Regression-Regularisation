**Réalisez une selection de features**

Que font les deux fonction ci-dessous ?

Testez les sur vos données, avec l'un de vos modèles. Quelle conclusion ?

```python
# ====================================================
# 
# ====================================================
def function1(model, X, y, cv=5, max_features=None):
    remaining_features = list(X.columns)
    selected_features = []
    best_score = -np.inf

    while remaining_features:
        if max_features and len(selected_features) >= max_features:
            break

        scores = []
        for feature in remaining_features:
            features_to_test = selected_features + [feature]
            score = cross_val_score(model, X[features_to_test], y, cv=cv, scoring='r2').mean()
            scores.append((score, feature))
        scores.sort(reverse=True)
        best_new_score, best_new_feature = scores[0]

        # On garde la variable si le score s’améliore
        if best_new_score > best_score + 1e-5:
            selected_features.append(best_new_feature)
            remaining_features.remove(best_new_feature)
            best_score = best_new_score
        else:
            break
    return selected_features, best_score

# ====================================================
# 
# ====================================================
def function2(model, X, y, cv=5, max_features=None):
    selected_features = list(X.columns)
    best_score = cross_val_score(model, X[selected_features], y, cv=cv, scoring='r2').mean()

    while len(selected_features) > max_features:
        

        scores = []
        for feature in selected_features:
            features_to_test = [f for f in selected_features if f != feature]
            score = cross_val_score(model, X[features_to_test], y, cv=cv, scoring='r2').mean()
            scores.append((score, feature))
        scores.sort(reverse=True)
        best_new_score, worst_feature = scores[0]
        selected_features.remove(worst_feature)

        if best_new_score > best_score + 1e-5:
            best_score = best_new_score
        
    return selected_features, best_score
```
